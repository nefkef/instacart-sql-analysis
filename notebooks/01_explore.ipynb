# Data exploration

## Goal
In this notebook we look at the data for each table to understand the contents and how they connect to each other.

## Tables
`aisles`, `departments`, `products`, `orders`, `order_products_prior`, `order_products_train`

## Key columns
- `reordered`: This value indicates whether a customer has ordered a product before. Reordered has two values, 0 or 1. 0 means no and 1 means yes.
- `days_since_prior_order`:  If "days since the prior order" has no value we see the value NaN on the table. This means that this is the first order of this user and we have no data of a previous one. We checked this with SQL : 206,209 orders were first orders. Every first order has no value, so every first order is flagged NULL. And every NULL row has `order_number` equal to 1, so every NULL is a first order.
- `order_dow`: This value shows us what day of the week the order was placed, with values from 0 to 6.

## Assumptions
- We assume that `order_dow` 0 = Sunday. The dataset does not document which day is 0. If this is wrong, our conclusions about weekend orders could be wrong.
- We assume that `order_hour_of_day` is the customer's local time. The dataset does not document the time zone, and the USA has several time zones. If this is wrong, our conclusions about the time of day people shop could be wrong.
